```python
class InsufficientFundsError(Exception):
    """Raised when withdrawal amount exceeds account balance."""
    pass


class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")
        self.balance += amount
        print(f"Deposited: GHS {amount:.2f}")

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")
        if amount > self.balance:
            raise InsufficientFundsError("Insufficient funds.")
        self.balance -= amount
        print(f"Withdrawn: GHS {amount:.2f}")

    def check_balance(self):
        print(f"Current Balance: GHS {self.balance:.2f}")


# Demonstration
account = BankAccount(1000)

try:
    account.check_balance()

    account.deposit(500)
    account.check_balance()

    account.withdraw(300)
    account.check_balance()

    # Uncomment these lines to test exceptions

    # account.deposit(-100)        # Negative deposit
    # account.withdraw(-50)        # Invalid withdrawal amount
    # account.withdraw(5000)       # Insufficient funds

except ValueError as e:
    print("Error:", e)

except InsufficientFundsError as e:
    print("Error:", e)
```
